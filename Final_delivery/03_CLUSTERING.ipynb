# 4. Clustering Analysis

### Correlation Matrix

```


# List of numerical columns
numerical_cols = [
    "trip_distance", "fare_amount", "extra", "mta_tax", "tip_amount",
    "tolls_amount", "improvement_surcharge", "total_amount",
    "congestion_surcharge", "pickup_hour_decimal",
    "trip_duration_minutes", "trip_speed_mph"
]

# Assemble into 'features' column
assembler = VectorAssembler(inputCols=numerical_cols, outputCol="features")
df_vector = assembler.transform(df.select(numerical_cols))

# Compute correlation matrix
correlation_matrix = Correlation.corr(df_vector, "features", method="pearson").head()[0]

# Convert matrix to numpy array
corr_array = correlation_matrix.toArray()

# Convert to pandas DataFrame for easier viewing
corr_df = pd.DataFrame(corr_array, columns=numerical_cols, index=numerical_cols)
corr_df
```

### Normalization 

```
scaler = StandardScaler(
    inputCol="features", 
    outputCol="scaled_features",
    withMean=True, 
    withStd=True
)

scaler_model = scaler.fit(df_vector)
df_scaled = scaler_model.transform(df_vector)
```

## Kmeans 

Since the Cluster can not run for all DataSet, we randomly choose 50.000 lines to experiment KMeans 

```

sample_df = df_scaled.orderBy(rand()).limit(50000)
```

#### Was made tests to find out the ideal K
If we had more computation capacity we would try for all dataset and the K would probably be differente

```

# Initialize KMeans with the desired number of clusters
kmeans = KMeans(featuresCol="scaled_features", k=3, seed=42)

# Fit the model
model = kmeans.fit(sample_df)

# Make predictions (assign clusters)
df_clusters = model.transform(sample_df)

# Show some results
df_clusters.select("scaled_features", "prediction").show(10, truncate=False)
```

```
centers = model.clusterCenters()
for idx, center in enumerate(centers):
    print(f"Cluster {idx}: {center}")
```

```
df_clusters.groupBy("prediction").count().orderBy("prediction").show()
```

```

# Apply PCA to reduce dimensions to 2
pca = PCA(k=2, inputCol="scaled_features", outputCol="pca_features")
pca_model = pca.fit(df_clusters)
df_pca = pca_model.transform(df_clusters)
```

```
# Select PCA columns and cluster prediction
plot_data = df_pca.select("pca_features", "prediction")

# Extract coordinates
plot_data = plot_data.rdd.map(lambda row: (float(row['pca_features'][0]), 
                                           float(row['pca_features'][1]), 
                                           row['prediction'])).toDF(["x", "y", "cluster"])

# Convert to Pandas for plotting
pandas_df = plot_data.toPandas()
```

```

plt.figure(figsize=(10, 6))
sns.scatterplot(data=pandas_df, x="x", y="y", hue="cluster", palette="Set2", s=50)
plt.title("Cluster visualization after PCA (K=3)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.legend(title="Cluster")
plt.show()
```

## DBScan 


### With PCA

```

# Convert Spark DataFrame to Pandas
sample_pd = sample_df.select("scaled_features") \
    .withColumn("features_array", vector_to_array("scaled_features")) \
    .select("features_array").toPandas()

# Convert to list of feature vectors
X = sample_pd["features_array"].tolist()

# Apply DBSCAN
dbscan = DBSCAN(eps=0.3, min_samples=5)
labels = dbscan.fit_predict(X)
sample_pd["cluster"] = labels

# Reduce dimensions to 2D using PCA
pca = PCA(n_components=2)
X_2d = pca.fit_transform(X)

# Create plot
plt.figure(figsize=(10, 6))
scatter = plt.scatter(X_2d[:, 0], X_2d[:, 1], c=labels, cmap='tab10', s=10)
plt.title("DBSCAN Clusters (PCA 2D)")
plt.xlabel("PCA 1")
plt.ylabel("PCA 2")
plt.colorbar(scatter, label='Cluster ID')
plt.grid(True)
plt.show()
```

This type of vizualization it's nor good for a dense and spare dataset

### With t-SNE

```

# Convert Spark DataFrame to Pandas
sample_pd = sample_df.select("scaled_features") \
    .withColumn("features_array", vector_to_array("scaled_features")) \
    .select("features_array").toPandas()

# Transform to feature matrix
X = sample_pd["features_array"].tolist()

# Run DBSCAN
dbscan = DBSCAN(eps=0.8, min_samples=15)
labels = dbscan.fit_predict(X)
sample_pd["cluster"] = labels

# Reduce dimensions with t-SNE
tsne = TSNE(n_components=2, random_state=42, perplexity=30, n_iter=1000)
X_tsne = tsne.fit_transform(X)

# Plot
plt.figure(figsize=(10, 6))
scatter = plt.scatter(X_tsne[:, 0], X_tsne[:, 1], c=labels, cmap='tab10', s=10)
plt.title("DBSCAN Clusters visualized with t-SNE (2D)")
plt.xlabel("t-SNE 1")
plt.ylabel("t-SNE 2")
plt.colorbar(scatter, label='Cluster ID')
plt.grid(True)
plt.show()
```
