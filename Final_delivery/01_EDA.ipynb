```python
# List for all the parquet files 
file_paths = [f.path for f in parquets if f.path.endswith(".parquet")]
```

```python
file_paths
```

```python
# Function to extract the month from the filename
def extract_year_month(filename):
    filename= filename.replace('.parquet','') # To remove .parquet

    # Get the file name from the full path
    file_name = filename.split('/')[-1]  # ex: 'yellow_tripdata_2024_01'

    # Then split the name
    name_split= file_name.split('_')

    # Get the year and month
    year= name_split[2]
    month= name_split[3]
    
    return year,month
```

```python
# Loads the files and adds the 'month' column
dfs = []
for path in file_paths:
    year, month = extract_year_month(path)
    df = spark.read.parquet(path)
    df = df.withColumn("month", F.lit(int(month)))  # adds the column month
    df= df.withColumn("year", F.lit(int(year))) 
    dfs.append(df)
```

```python
# Joins everything into a single DataFrame
full_df = reduce(DataFrame.unionByName, dfs)
```

```python
print("\nNull values per column (%):")

total_rows = full_df.count()

for column in full_df.columns:
    nulls = full_df.filter(F.col(column).isNull()).count()
    if nulls > 0:
        percentage = (nulls / total_rows) * 100
        print(f"{column}: {nulls} nulls ({percentage:.2f}%)")
```

The consistent percentage across multiple fields suggests that these null values ​​likely originate from the same subset of rows, possibly due to:
- A specific provider or taxi type not reporting these fields
- A data ingestion issue

Let's check if there are actually nulls originating from the same subset of rows:

```python
same_nulls = full_df.filter(
    F.col("passenger_count").isNull() & 
    F.col("RatecodeID").isNull() &
    F.col("store_and_fwd_flag").isNull() &
    F.col("congestion_surcharge").isNull() &
    F.col("Airport_fee").isNull()
)

same_nulls.count()
```

```python
same_nulls.limit(100).toPandas()
```

```python
# Total number of records per VendorID in the full dataset
total_vendor = full_df.groupBy("VendorID").count().withColumnRenamed("count", "total_rows")

# Total number of records with nulls per VendorID
nulls_vendor = same_nulls.groupBy("VendorID").count().withColumnRenamed("count", "null_rows")

# Join both and calculate the percentage
vendor_null_stats = total_vendor.join(nulls_vendor, on="VendorID", how="left") \
    .withColumn("null_rows", F.coalesce("null_rows", F.lit(0))) \
    .withColumn("null_percentage", (F.col("null_rows") / F.col("total_rows")) * 100)

# Display the result
vendor_null_stats.select(
    "VendorID", 
    "total_rows", 
    "null_rows", 
    F.round("null_percentage", 2).alias("null_percentage")
).orderBy("VendorID").show()
```

Upon further investigation, it was confirmed that all missing values ​​occur in the same subset of rows.
However, this subset is spread across multiple months, years, and VendorIDs, indicating that the missing values are not caused by a single corrupt batch or a vendor-specific reporting issue, but are more likely the result of systemic and inconsistent data collection practices.

🔧 __Preprocessing Decision__:
Given that the affected rows:
- Represent only 5.68% of the dataset;
- Are not biased toward a specific time period or vendor;
- And contain missing values in multiple critical columns. <br>
We decided to drop these rows on preprocessing. Retaining them would introduce noise and unreliability into downstream analysis

## Categorical vs Categorical

```python
categorical= ["VendorID","RatecodeID","store_and_fwd_flag","payment_type","PULocationID","DOLocationID"]
```

```python
cat_cols = [column for column in categorical if sample_df[column].nunique() <= 20]
```

```python
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    phi2_corr = max(0, phi2 - ((k-1)*(r-1))/(n-1))
    r_corr = r - ((r-1)**2)/(n-1)
    k_corr = k - ((k-1)**2)/(n-1)
    return np.sqrt(phi2_corr / min((k_corr-1), (r_corr-1)))
```

```python
cramer_matrix = pd.DataFrame(index=cat_cols, columns=cat_cols)

for col1 in cat_cols:
    for col2 in cat_cols:
        if col1 == col2:
            cramer_matrix.loc[col1, col2] = 1.0
        else:
            try:
                cramer_matrix.loc[col1, col2] = cramers_v(sample_df[col1], sample_df[col2])
            except:
                cramer_matrix.loc[col1, col2] = np.nan

cramer_matrix = cramer_matrix.astype(float)

# Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(cramer_matrix, annot=True, cmap="YlGnBu", fmt=".2f", square=True)
plt.title("Cramér’s V Association Between Categorical Variables")
plt.tight_layout()
plt.show()
```

##### 📘 Interpretation: Cramér’s V Association Between Categorical Variables

Cramér’s V is designed to detect association patterns between categorical variables, and low values here indicate no strong relationship between them.

The heatmap shows that the categorical variables have **no significant association** with each other — as expected, the diagonal contains `1.00` values (each variable is perfectly associated with itself), while all off-diagonal values are near zero.

This result suggests that variables are **statistically independent** in this dataset. The values of one variable do **not influence or depend** on the values of another.

## Numerical vs Categorical

```python
def eta_squared(df, num_col, cat_col):
    # Keep only the relevant columns and drop missing values
    df = df[[num_col, cat_col]].dropna()

    # Convert the categorical column to a proper category type
    categories = df[cat_col].astype("category")

    # Extract the numerical column (the dependent variable)
    y = df[num_col]

    # Create a list of numerical values grouped by category
    groups = [y[categories == cat] for cat in categories.cat.categories]

    # Compute the Between-Group Sum of Squares (SS_between)
    ss_between = sum(len(g) * (g.mean() - y.mean())**2 for g in groups)

    # Compute the Total Sum of Squares (SS_total)
    ss_total = sum((y - y.mean())**2)

    # Return Eta Squared: the proportion of variance explained by the categorical variable
    return ss_between / ss_total if ss_total != 0 else np.nan
```

```python
results = []

for cat in categorical:
    for num in numeric_cols:
        try:
            score = eta_squared(sample_df.dropna(subset=[cat, num]), num, cat)
            results.append((cat, num, round(score, 3)))
        except:
            results.append((cat, num, None))

eta_df = pd.DataFrame(results, columns=["Categorical", "Numerical", "Eta²"])
display(eta_df)
```

```python
eta_df["Eta²"] = pd.to_numeric(eta_df["Eta²"], errors="coerce")

eta_pivot = eta_df.pivot(index="Categorical", columns="Numerical", values="Eta²")

print(eta_pivot)

plt.figure(figsize=(10, 6))
sns.heatmap(eta_pivot, annot=True, cmap="YlGnBu", fmt=".2f", linewidths=0.5, mask=eta_pivot.isnull())
plt.title("Eta Squared (η²) – Association Between Categorical and Numerical Variables")
plt.tight_layout()
plt.show()
```

##### 📊 Interpretation of Categorical vs Numerical Association (Eta Squared)

The heatmap above visualizes the strength of association between categorical and numerical variables using Eta Squared. This metric quantifies the proportion of variance in a numerical variable that is explained by a categorical variable, ranging from 0 (no association) to 1 (perfect association).

###### 🔍 Key Insights:

- **PULocationID** shows a **strong association with `airport_fee`** (η² = 0.83), suggesting that certain pickup locations are highly indicative of whether an airport fee was applied.
- **RatecodeID** has moderate associations with multiple variables, such as:
  - `fare_amount` (η² = 0.34)
  - `tip_amount` and `total_amount` (η² = 0.35)
  These results make sense, as different rate codes often imply different fare structures and conditions.
- **VendorID** only shows a notable association with `extra` charges (η² = 0.33), which may reflect differences in vendor-specific policies or fees.
- **payment_type** moderately explains `mta_tax` (η² = 0.27), possibly due to differences in tax handling between cash and card payments.
- `store_and_fwd_flag` has negligible influence across all numerical variables, indicating it is not a meaningful segmenter for quantitative analysis in this dataset.
